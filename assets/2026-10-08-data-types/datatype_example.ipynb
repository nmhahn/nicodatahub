# Libraries

```python
import pandas as pd
```

# Dataset

- **Name**: *Individual Household Electric Power Consumption*
- **Source**: https://archive.ics.uci.edu/dataset/235/individual+household+electric+power+consumption

```python
df = pd.read_table("household_power_consumption.txt", sep=";", decimal=",", low_memory=False)
print(f"   Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
```

In order to ease the illustration and manipulation I'll do a few preprocessing on this dataset:
- lowercase all column names;
- combine `Date` and `Time` in a single `DateTime` column;
- create `Year` and `Month` columns.

```python
df.columns = [col.lower() for col in df.columns]
df.columns
```

```python
df["datetime"] = df["date"]+" "+df["time"]
df.drop(columns=["date","time"], inplace=True)
```

```python
df["datetime"] = pd.to_datetime(df["datetime"], format="%d/%m/%Y %H:%M:%S")
df["year"] = df["datetime"].dt.year.astype("str")
df["month"] = df["datetime"].dt.month_name()

df["datetime"] = df["datetime"].dt.strftime("%d/%m/%Y %H:%M:%S")
```

```python
df = df[ list(df.columns[-3:]) + list(df.columns[:-3]) ] # just reordering
df.head()
```

# Dataset Memory Consumption

```python
# Notice that all columns are stored as object/strings
df.dtypes
```

```python
memory = df.memory_usage(deep=True)

print(round(memory / (1024 ** 2), 2))
print()

memory_old = round(memory.sum() / (1024 ** 2), 2)
print(f"Dataset Memory Consumption: {memory_old} MB") 
```

```python
df["datetime"] = pd.to_datetime(df["datetime"], format="%d/%m/%Y %H:%M:%S")
df["year"] = df["year"].astype("int")
df["month"] = df["month"].astype("category")
df.head()
```

```python
df["global_active_power"] = df["global_active_power"].replace("?",None).astype("float")
df["global_reactive_power"] = df["global_reactive_power"].replace("?",None).astype("float")
df["voltage"] = df["voltage"].replace("?",None).astype("float")
df["global_intensity"] = df["global_intensity"].replace("?",None).astype("float")
df["sub_metering_1"] = df["sub_metering_1"].replace("?",None).astype("float")
df["sub_metering_2"] = df["sub_metering_2"].replace("?",None).astype("float")
df["sub_metering_3"] = df["sub_metering_3"].replace("?",None).astype("float")
```

```python
# Now all columns have proper data types
df.dtypes
```

```python
memory = df.memory_usage(deep=True)

print(round(memory / (1024 ** 2), 2))
print()

memory_now = round(memory.sum() / (1024 ** 2), 2)
print(f"Dataset Memory Consumption: {memory_now} MB") 
```

```python
# Same data, proper data types, less memory usage
print(f"Dataset Memory Consumption (before): {memory_old} MB") 
print(f"Dataset Memory Consumption  (after):  {memory_now} MB")
```
